# Abrir Roadkill desde Jupyter

Ejecuta **Kernel → Reiniciar y ejecutar todas las celdas** (o las celdas 1, 2 y 3 por orden). La aplicación se abrirá en tu navegador y el notebook quedará disponible.

Coloca este archivo en la **carpeta principal del ZIP descomprimido**, junto a `requirements.txt`, `backend/`, `frontend/` y `data/`. También puede estar en una subcarpeta de esa aplicación. No lo ejecutes dentro del ZIP.

Necesitas Jupyter con un kernel local de **Python 3.10 o superior**. La primera instalación requiere internet y puede tardar unos minutos. No necesitas ejecutar los `.bat`, instalar Node.js ni abrir el servidor de datasets: esta aplicación utiliza sus resultados guardados. Los mapas de fondo pueden necesitar internet.

El notebook crea un entorno Python propio `.venv_jupyter`, por lo que las dependencias de la aplicación no se instalan en el kernel de tus otros notebooks. Puedes compartir este archivo sin modificar rutas personales. En Jupyter remoto/Colab la apertura del navegador local y el acceso a `127.0.0.1` no funcionan como en tu ordenador.


## 1. Localizar la aplicación

Normalmente no cambies nada. Si Jupyter trabaja desde otra carpeta y no encuentra la aplicación, escribe en `CARPETA_APP` su ruta, que puede ser relativa al directorio de trabajo de Jupyter.


```python
from pathlib import Path
import atexit
import hashlib
import json
import os
import socket
import subprocess
import sys
import time
import urllib.request
import webbrowser
from IPython.display import display, HTML

CARPETA_APP = None  # Ejemplo relativo: "../ROADKILL_V15_INICIO_JUPYTER"
AADF_CSV_OVERRIDE = None  # Ruta al original, solo si está fuera de la aplicación y sus carpetas superiores.
PUERTO_PREFERIDO = 8000
ABRIR_NAVEGADOR = True

if sys.version_info < (3, 10):
    raise RuntimeError("Selecciona un kernel de Python 3.10 o superior.")

def es_aplicacion(path):
    return all((path / p).is_file() for p in (
        "requirements.txt", "backend/app.py", "frontend/dist/index.html",
        "data/predicciones_test_2017.csv",
    ))

if CARPETA_APP is not None:
    candidatos = [Path(CARPETA_APP).expanduser().resolve()]
else:
    inicio = Path.cwd().resolve()
    candidatos = [inicio, *inicio.parents]

RAIZ_APP = next((p for p in candidatos if es_aplicacion(p)), None)
if RAIZ_APP is None:
    raise FileNotFoundError(
        "No encuentro la aplicación descomprimida. Guarda este notebook junto a "
        "requirements.txt, backend/, frontend/ y data/. Si Jupyter utiliza otro "
        "directorio, indica CARPETA_APP y vuelve a ejecutar esta celda. "
        f"Directorio de trabajo: {Path.cwd()}"
    )

DIR_LOGS = RAIZ_APP / "logs_jupyter"
DIR_LOGS.mkdir(exist_ok=True)
LOG_SERVIDOR = DIR_LOGS / "servidor.log"
LOG_INSTALACION = DIR_LOGS / "instalacion.log"
print("Aplicación localizada:", RAIZ_APP)
print("Python del kernel:", sys.version.split()[0])
print("No se ejecutará ningún reentrenamiento.")
```

## 2. Preparar las dependencias

La primera ejecución crea el entorno e instala los paquetes de `requirements.txt`. Las siguientes reutilizan esa instalación. Si cambia el archivo de dependencias, se vuelve a instalar automáticamente. Si falta algún paquete o la instalación está dañada, se intenta repararla.


```python
ENTORNO = RAIZ_APP / ".venv_jupyter"
PYTHON_APP = ENTORNO / ("Scripts/python.exe" if os.name == "nt" else "bin/python")


def ejecutar_preparacion(args):
    with LOG_INSTALACION.open("a", encoding="utf-8") as log:
        log.write("\n" + " ".join(map(str, args)) + "\n")
        log.flush()
        resultado = subprocess.run(
            list(map(str, args)), cwd=str(RAIZ_APP),
            stdout=log, stderr=subprocess.STDOUT, timeout=900,
        )
    if resultado.returncode:
        detalle = LOG_INSTALACION.read_text(encoding="utf-8", errors="replace")[-5000:]
        raise RuntimeError(
            "No se pudo preparar el entorno. Revisa tu conexión y el kernel de Python. "
            f"Registro: {LOG_INSTALACION}\n\n{detalle}"
        )


if not PYTHON_APP.is_file():
    print("Creando el entorno de la aplicación…")
    ejecutar_preparacion([sys.executable, "-m", "venv", str(ENTORNO)])

REQUISITOS = RAIZ_APP / "requirements.txt"
MARCADOR = ENTORNO / "roadkill_requirements.sha256"
huella = hashlib.sha256(REQUISITOS.read_bytes()).hexdigest()
prueba_imports = subprocess.run(
    [str(PYTHON_APP), "-c",
     "import fastapi, uvicorn, pandas, numpy, pyproj, sklearn, astral"],
    cwd=str(RAIZ_APP), capture_output=True, text=True,
)
instalado = MARCADOR.is_file() and MARCADOR.read_text().strip() == huella
if not instalado or prueba_imports.returncode:
    print("Instalando dependencias; la primera vez puede tardar unos minutos…")
    ejecutar_preparacion([
        str(PYTHON_APP), "-m", "pip", "install", "--disable-pip-version-check",
        "-r", str(REQUISITOS),
    ])
    ejecutar_preparacion([
        str(PYTHON_APP), "-c",
        "import fastapi, uvicorn, pandas, numpy, pyproj, sklearn, astral",
    ])
    MARCADOR.write_text(huella, encoding="utf-8")
else:
    print("Dependencias disponibles; no hace falta volver a instalarlas.")
print("Entorno preparado.")
```

## Preparar aforos de 2017

Si se encuentra el AADF original, se preparan medias específicas por carretera y celda. Si falta, la demo abre usando el perfil histórico 2008–2017 incluido en el ZIP. La pantalla identifica el origen y permite seleccionar el modo estricto AADF 2017.


```python
args_afluencia = [str(PYTHON_APP), str(RAIZ_APP / "scripts/prepare_route_traffic.py")]
if AADF_CSV_OVERRIDE is not None:
    args_afluencia += ["--aadf", str(Path(AADF_CSV_OVERRIDE).expanduser().resolve())]
if (RAIZ_APP / "data/route_traffic_2017.csv").is_file() and AADF_CSV_OVERRIDE is None:
    print("Afluencia: se reutiliza la exportación AADF 2017 de data/route_traffic_2017.csv.")
else:
    preparado = subprocess.run(args_afluencia, cwd=RAIZ_APP, capture_output=True, text=True)
    if preparado.returncode != 0:
        raise RuntimeError("No se pudo preparar la afluencia AADF. Revisa el original:\n" + preparado.stderr[-3000:])
    print(preparado.stdout)
```

## 3. Arrancar y abrir el frontend

La celda devuelve el control al notebook cuando el backend responde. Si vuelves a ejecutarla en el mismo kernel, reutiliza el proceso de esta aplicación. Si el puerto 8000 está ocupado, utiliza otro disponible y muestra su dirección. No detiene otros servidores.

Deja Jupyter abierto mientras uses la aplicación. Para detenerla antes de cerrar o reiniciar el kernel, utiliza la última celda. Un reinicio forzado del kernel podría dejar el servidor anterior funcionando; si ocurre, la siguiente ejecución elegirá un puerto libre.


```python
# Conservamos las referencias al repetir celdas en el mismo kernel.
if "_roadkill_procesos" not in globals():
    _roadkill_procesos = []


def detener_app():
    """Detiene solo los procesos iniciados por este notebook en este kernel."""
    for proceso in _roadkill_procesos:
        if proceso.poll() is None:
            proceso.terminate()
            try:
                proceso.wait(timeout=8)
            except subprocess.TimeoutExpired:
                proceso.kill()
                proceso.wait(timeout=8)
    print("Aplicaciones iniciadas por este notebook detenidas.")


if not globals().get("_roadkill_salida_registrada", False):
    atexit.register(detener_app)
    _roadkill_salida_registrada = True


def seleccionar_puerto(preferido):
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as sock:
        try:
            sock.bind(("127.0.0.1", preferido))
        except OSError:
            sock.bind(("127.0.0.1", 0))
        return sock.getsockname()[1]


def pedir_json(url):
    # La conexión al backend local no utiliza proxies del sistema.
    cliente = urllib.request.build_opener(urllib.request.ProxyHandler({}))
    with cliente.open(url, timeout=2) as respuesta:
        return json.load(respuesta)


def firma_aplicacion():
    """Detecta cambios de código y datos para no reutilizar un servidor antiguo."""
    digest = hashlib.sha256()
    archivos = list((RAIZ_APP / "backend").glob("*.py"))
    archivos += [p for p in (RAIZ_APP / "frontend/dist").rglob("*") if p.is_file()]
    archivos += [RAIZ_APP / "data" / nombre for nombre in (
        "predicciones_test_2017.csv", "cell_coordinates_reto06.csv", "route_traffic_2017.csv",
    ) if (RAIZ_APP / "data" / nombre).is_file()]
    for archivo in sorted(archivos):
        digest.update(str(archivo.relative_to(RAIZ_APP)).encode("utf-8"))
        digest.update(archivo.read_bytes())
    return digest.hexdigest()

firma_actual = firma_aplicacion()
proceso_propio_vivo = (
    globals().get("_roadkill_proceso") is not None
    and _roadkill_proceso.poll() is None
    and globals().get("_roadkill_raiz_proceso") == str(RAIZ_APP)
)
if proceso_propio_vivo and globals().get("_roadkill_firma_proceso") != firma_actual:
    print("Cambios detectados: reiniciando el servidor iniciado por este notebook.")
    _roadkill_proceso.terminate()
    try:
        _roadkill_proceso.wait(timeout=8)
    except subprocess.TimeoutExpired:
        _roadkill_proceso.kill()
        _roadkill_proceso.wait(timeout=8)
    proceso_propio_vivo = False
vivo = proceso_propio_vivo
if vivo:
    URL_APP = _roadkill_url
    print("Reutilizando la aplicación ya iniciada.")
else:
    puerto = seleccionar_puerto(int(PUERTO_PREFERIDO))
    URL_APP = f"http://127.0.0.1:{puerto}"
    entorno_hijo = os.environ.copy()
    entorno_hijo["PYTHONUNBUFFERED"] = "1"
    entorno_hijo.pop("PYTHONPATH", None)
    entorno_hijo.pop("PYTHONHOME", None)
    with LOG_SERVIDOR.open("a", encoding="utf-8") as log:
        log.write(f"\nIniciando {URL_APP} desde {RAIZ_APP}\n")
        log.flush()
        _roadkill_proceso = subprocess.Popen(
            [str(PYTHON_APP), "-m", "uvicorn", "backend.app:app",
             "--host", "127.0.0.1", "--port", str(puerto)],
            cwd=str(RAIZ_APP), env=entorno_hijo,
            stdout=log, stderr=subprocess.STDOUT,
        )
    _roadkill_procesos.append(_roadkill_proceso)
    _roadkill_raiz_proceso = str(RAIZ_APP)
    _roadkill_url = URL_APP
    _roadkill_firma_proceso = firma_actual

print("Esperando a que responda el backend…")
listo = False
ultimo_error = ""
for _ in range(90):
    if _roadkill_proceso.poll() is not None:
        ultimo_error = "El proceso del servidor terminó antes de estar listo."
        break
    try:
        salud = pedir_json(URL_APP + "/api/health")
        listo = salud.get("status") == "ok" and salud.get("rows", 0) > 0
        if listo:
            break
    except Exception as exc:
        ultimo_error = str(exc)
    time.sleep(0.5)

if not listo:
    if _roadkill_proceso.poll() is None:
        _roadkill_proceso.terminate()
        try:
            _roadkill_proceso.wait(timeout=8)
        except subprocess.TimeoutExpired:
            _roadkill_proceso.kill()
            _roadkill_proceso.wait(timeout=8)
    detalle = LOG_SERVIDOR.read_text(encoding="utf-8", errors="replace")[-5000:]
    raise RuntimeError(
        f"La aplicación no ha arrancado. {ultimo_error}\n"
        f"Registro: {LOG_SERVIDOR}\n\n{detalle}"
    )

print(f"Aplicación disponible: {URL_APP}")
print(f"Backend comprobado: {salud['rows']:,} filas de predicciones.")
display(HTML(f'<p><a href="{URL_APP}" target="_blank" rel="noopener">Abrir Roadkill en el navegador</a></p>'))
if ABRIR_NAVEGADOR:
    try:
        abierto = webbrowser.open(URL_APP, new=2)
        if not abierto:
            print("Si no se abre automáticamente, pulsa el enlace anterior.")
    except Exception:
        print("Pulsa el enlace anterior para abrir la aplicación.")
```

## 4. Detener la aplicación (opcional)

Esta celda no detiene la aplicación al ejecutar todo el notebook. Cuando quieras cerrarla, cambia `DETENER` a `True` y ejecuta **solo esta celda**. Después, vuelve a dejarlo en `False`. También puedes ejecutar `detener_app()` en una celda nueva.


```python
DETENER = False
if DETENER:
    detener_app()
else:
    print("La aplicación permanece abierta. Para cerrarla: detener_app().")
```
